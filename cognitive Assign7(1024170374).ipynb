```python
import numpy as np
import pandas as pd

# Part I & II: Randomized Sales Data Generation & Analysis
np.random.seed(1024170374)
sales_data = np.random.randint(1000, 5001, size=(12, 4))
columns = ['Electronics', 'Clothing', 'Home & Kitchen', 'Sports']
index = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

df = pd.DataFrame(sales_data, columns=columns, index=index)
print("First 5 Rows:")
print(df.head())
print("\nSummary Statistics:")
print(df.describe())

df['Total Sales'] = df.sum(axis=1)
df['Growth Rate'] = df['Total Sales'].pct_change() * 100
# Roll number 1024170374 is even -> 10% discount on Electronics
df['Electronics_Discounted'] = df['Electronics'] * 0.90
print("\nModified DataFrame with Total Sales and Discounts:")
print(df[['Total Sales', 'Growth Rate', 'Electronics_Discounted']].head())
```
