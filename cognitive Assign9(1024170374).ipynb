```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, LancasterStemmer, WordNetLemmatizer

text = "Artificial intelligence and machine learning are revolutionizing software development in 2026. Contact support@ai.org or visit https://ai.org or call +91 9876543210 for details. State-of-the-art models handle complex tasks seamlessly."

# Q1 & Q2: Tokenization, Stemming, Lemmatization
words = re.findall(r'\b\w+\b', text.lower())
ps = PorterStemmer()
ls = LancasterStemmer()
lem = WordNetLemmatizer()
print("Porter:", [ps.stem(w) for w in words[:5]])
print("Lancaster:", [ls.stem(w) for w in words[:5]])
print("Lemmatized:", [lem.lemmatize(w) for w in words[:5]])
```

```python
# Q3 & Q4: Regex Extractions and Cleaning
emails = re.findall(r'[\w\.-]+@[\w\.-]+', text)
urls = re.findall(r'https?://\S+', text)
phones = re.findall(r'\+?\d[\d -]{8,12}\d', text)
cleaned_text = re.sub(r'[\w\.-]+@[\w\.-]+', '<EMAIL>', text)
cleaned_text = re.sub(r'https?://\S+', '<URL>', cleaned_text)
cleaned_text = re.sub(r'\+?\d[\d -]{8,12}\d', '<PHONE>', cleaned_text)
print("Cleaned Text:", cleaned_text)
```
