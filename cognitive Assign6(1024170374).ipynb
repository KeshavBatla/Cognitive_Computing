```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Q1: Mathematical functions plot
M = 2
x = np.linspace(-10, 10, 100)
y1 = M * (x ** 2)
y2 = M * np.sin(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y1, label='y = M*x^2', color='blue', linestyle='--')
plt.plot(x, y2, label='y = M*sin(x)', color='red', linestyle='-')
plt.title('Mathematical Functions Plot')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Q3: Random Seed Subplots
np.random.seed(1024170374)
data = np.random.randn(50)
cum_sum = np.cumsum(data)
noise = np.random.randn(50)

fig, axs = plt.subplots(2, 2, figsize=(10, 8))
axs[0, 0].plot(cum_sum, color='purple')
axs[0, 0].set_title('Cumulative Sum')
axs[0, 0].grid(True)

axs[0, 1].scatter(range(50), noise, color='green')
axs[0, 1].set_title('Scatter Plot with Noise')
axs[0, 1].grid(True)

plt.tight_layout()
plt.show()
```
