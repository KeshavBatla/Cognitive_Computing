```python
import numpy as np

# Q1: Basic 1D array operations
arr1d = np.array([1, 2, 3, 4, 5])
print("Original:", arr1d)
print("Add 2:", arr1d + 2)
print("Multiply 3:", arr1d * 3)
print("Divide 2:", arr1d / 2)
```

```python
# Q2: Reversing and Frequency
arr_rev = np.array([1, 2, 3, 6, 4, 5])
print("Reversed:", arr_rev[::-1])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
vals, counts = np.unique(x, return_counts=True)
most_freq_x = vals[np.argmax(counts)]
print("Most frequent in x:", most_freq_x, "Count:", np.max(counts))
```

```python
# Q3 & Q4 & Q5
arr2d = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print("1st row, 2nd col:", arr2d[0, 1])
print("3rd row, 1st col:", arr2d[2, 0])

keshav = np.linspace(10, 100, 25)
print("Keshav array shape:", keshav.shape, "Size:", keshav.size, "Bytes:", keshav.nbytes)

ucs420_keshav = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]).reshape(3, 4)
print("Mean:", np.mean(ucs420_keshav))
print("Median:", np.median(ucs420_keshav))
print("Max:", np.max(ucs420_keshav))
print("Min:", np.min(ucs420_keshav))
print("Unique:", np.unique(ucs420_keshav))
reshaped_ucs420_keshav = ucs420_keshav.reshape(4, 3)
print("Reshaped (4x3):\n", reshaped_ucs420_keshav)
resized_ucs420_keshav = np.resize(ucs420_keshav, (2, 3))
print("Resized (2x3):\n", resized_ucs420_keshav)
```
