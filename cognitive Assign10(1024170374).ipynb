```python
import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)

# Q1: Paragraph Processing & Frequency Distribution
text_q1 = "Artificial intelligence is transforming technology rapidly. Machine learning and deep learning models process vast amounts of data every second. Natural language processing empowers systems to understand human communication seamlessly. Cloud computing provides scalable infrastructure for modern applications. Innovation drives the future of digital engineering."
clean_text = re.sub(r'[^\w\s]', '', text_q1.lower())
words_split = clean_text.split()
words_nltk = word_tokenize(clean_text)
sentences = sent_tokenize(text_q1)
stop_words = set(stopwords.words('english'))
filtered_words = [w for w in words_nltk if w not in stop_words]
freq_dist = nltk.FreqDist(filtered_words)
print("Q1 Freq Dist:", freq_dist.most_common(5))
```

```python
# Q2: Stemming vs Lemmatization
alpha_words = [w for w in words_nltk if w.isalpha() and w not in stop_words]
ps = PorterStemmer()
lemmatizer = WordNetLemmatizer()
stemmed = [ps.stem(w) for w in alpha_words]
lemmatized = [lemmatizer.lemmatize(w) for w in alpha_words]
print("Stemmed:", stemmed[:5])
print("Lemmatized:", lemmatized[:5])
```

```python
# Q3: BoW and TF-IDF
texts = ["Cloud computing enables scalable backend services.", "Machine learning powers intelligent cognitive systems.", "Data analytics drives modern business decisions."]
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(texts)
print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
```
